# 1. Right Angled Pattern Program



```python
num = int(input("Enter the number: "))
for i in range(num):
    for j in range(i + 1):
        print("*", end=" ")
    print()
```

# 2. Inverted Right Angled Triangle

```python
num = int(input('Enter the number: '))
print('Enter the number: 7')

for i in range(num - 1, -1, -1):
    for j in range(i + 1):
        print('*', end=' ')
    print()
```
